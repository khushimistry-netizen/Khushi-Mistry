```python
# ============================================
# STOCK PORTFOLIO TRACKER
# ============================================

# Dictionary containing stock prices
stock_prices = {
    "AAPL": 180,
    "TSLA": 250,
    "GOOGL": 140,
    "MSFT": 420,
    "AMZN": 185,
    "META": 500,
    "NFLX": 680,
    "NVDA": 120,
    "AMD": 160,
    "INTC": 25
}

# Dictionary to store user's portfolio
portfolio = {}

print("=" * 50)
print("       STOCK PORTFOLIO TRACKER")
print("=" * 50)

print("\nAvailable Stocks:")
for stock, price in stock_prices.items():
    print(stock, "-> $", price)

print("\nEnter your stock details.")
print("Type 'done' when you have finished.\n")


# --------------------------------------------
# Taking input from user
# --------------------------------------------

while True:

    stock_name = input("Enter stock name: ").upper()

    # Stop taking input
    if stock_name == "DONE":
        break

    # Check whether stock exists
    if stock_name not in stock_prices:
        print("Stock not available.")
        print("Please choose from the available stocks.\n")
        continue

    # Take quantity
    try:
        quantity = int(input("Enter quantity: "))

        if quantity <= 0:
            print("Quantity must be greater than 0.\n")
            continue

    except ValueError:
        print("Please enter a valid number.\n")
        continue

    # Store stock quantity
    if stock_name in portfolio:
        portfolio[stock_name] += quantity
    else:
        portfolio[stock_name] = quantity

    print(quantity, stock_name, "added to portfolio.\n")


# --------------------------------------------
# Display Portfolio
# --------------------------------------------

print("\n" + "=" * 60)
print("                 YOUR PORTFOLIO")
print("=" * 60)

print(f"{'Stock':<10}{'Quantity':<12}{'Price':<12}{'Investment':<15}")
print("-" * 60)


# Total investment
total_investment = 0

# Store results for file
portfolio_data = []

for stock, quantity in portfolio.items():

    price = stock_prices[stock]

    investment = quantity * price

    total_investment += investment

    print(
        f"{stock:<10}"
        f"{quantity:<12}"
        f"${price:<11}"
        f"${investment:<15}"
    )

    portfolio_data.append(
        [stock, quantity, price, investment]
    )


# --------------------------------------------
# Display Total
# --------------------------------------------

print("-" * 60)

print("Total Investment: $", total_investment)

print("=" * 60)


# --------------------------------------------
# Additional Information
# --------------------------------------------

number_of_stocks = len(portfolio)

total_quantity = sum(portfolio.values())

print("\nPortfolio Summary")
print("-" * 30)

print("Different Stocks :", number_of_stocks)
print("Total Shares     :", total_quantity)
print("Total Investment :", "$", total_investment)


# --------------------------------------------
# Save portfolio in TXT file
# --------------------------------------------

save_txt = input(
    "\nDo you want to save the portfolio as TXT? (yes/no): "
).lower()

if save_txt == "yes":

    with open("portfolio.txt", "w") as file:

        file.write("STOCK PORTFOLIO REPORT\n")
        file.write("=" * 50 + "\n")

        file.write(
            f"{'Stock':<10}"
            f"{'Quantity':<12}"
            f"{'Price':<12}"
            f"{'Investment':<15}\n"
        )

        file.write("-" * 50 + "\n")

        for data in portfolio_data:

            stock = data[0]
            quantity = data[1]
            price = data[2]
            investment = data[3]

            file.write(
                f"{stock:<10}"
                f"{quantity:<12}"
                f"${price:<11}"
                f"${investment:<15}\n"
            )

        file.write("-" * 50 + "\n")

        file.write(
            f"Total Investment: ${total_investment}\n"
        )

    print("Portfolio saved successfully in portfolio.txt")


# --------------------------------------------
# Save portfolio in CSV file
# --------------------------------------------

save_csv = input(
    "\nDo you want to save the portfolio as CSV? (yes/no): "
).lower()

if save_csv == "yes":

    import csv

    with open(
        "portfolio.csv",
        "w",
        newline=""
    ) as file:

        writer = csv.writer(file)

        # Header
        writer.writerow(
            ["Stock", "Quantity", "Price", "Investment"]
        )

        # Data
        for data in portfolio_data:
            writer.writerow(data)

        # Total
        writer.writerow([])
        writer.writerow(
            ["Total Investment", "", "", total_investment]
        )

    print("Portfolio saved successfully in portfolio.csv")


# --------------------------------------------
# Final Message
# --------------------------------------------

print("\n" + "=" * 50)
print("       PORTFOLIO TRACKER COMPLETED")
print("=" * 50)

print("Thank you for using Stock Portfolio Tracker!")

print("\nFinal Investment Value: $", total_investment)

print("=" * 50)
```
