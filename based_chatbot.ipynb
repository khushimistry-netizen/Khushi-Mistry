```python


print("================================================")
print("           WELCOME TO MY CHATBOT")
print("================================================")
print("Type 'bye' to exit the chatbot.")
print("You can ask me simple questions.")
print("================================================")



def greeting():
    print("Bot: Hi! Nice to meet you.")



def how_are_you():
    print("Bot: I'm fine, thanks!")
    print("Bot: How are you?")



def good_mood():
    print("Bot: That's great to hear!")
    print("Bot: I am happy that you are doing well.")



def bad_mood():
    print("Bot: I'm sorry to hear that.")
    print("Bot: I hope your day gets better.")



def introduce():
    print("Bot: I am a simple rule-based chatbot.")
    print("Bot: I answer questions using predefined rules.")



def ask_name():
    print("Bot: My name is BasicBot.")
    print("Bot: You can call me Bot.")


def help_user():
    print("Bot: You can type the following:")
    print("Bot: hello")
    print("Bot: hi")
    print("Bot: how are you")
    print("Bot: what is your name")
    print("Bot: who are you")
    print("Bot: what can you do")
    print("Bot: thank you")
    print("Bot: bye")



def thank_you():
    print("Bot: You're welcome!")
    print("Bot: I'm happy to help you.")



def good_morning():
    print("Bot: Good morning!")
    print("Bot: Have a wonderful day.")


def good_afternoon():
    print("Bot: Good afternoon!")
    print("Bot: I hope you are having a good day.")



def good_evening():
    print("Bot: Good evening!")
    print("Bot: How was your day?")



def study():
    print("Bot: Keep studying regularly.")
    print("Bot: Practice makes programming easier.")



def python():
    print("Bot: Python is a popular programming language.")
    print("Bot: It is easy to learn and widely used in AI and Data Science.")


def artificial_intelligence():
    print("Bot: AI stands for Artificial Intelligence.")
    print("Bot: AI allows machines to perform intelligent tasks.")


def data_science():
    print("Bot: Data Science deals with collecting and analyzing data.")
    print("Bot: Python, Pandas and NumPy are commonly used.")



def chatbot_info():
    print("Bot: A chatbot is a program that communicates with users.")
    print("Bot: This chatbot uses simple if-elif rules.")



def weather():
    print("Bot: I cannot check live weather.")
    print("Bot: But I hope the weather is pleasant today!")


def joke():
    print("Bot: Why did the computer go to the doctor?")
    print("Bot: Because it had a virus!")



def motivation():
    print("Bot: Believe in yourself.")
    print("Bot: Keep learning and keep practicing!")
    print("Bot: Every expert was once a beginner.")


def unknown():
    print("Bot: Sorry, I don't understand that.")
    print("Bot: Type 'help' to see what I can answer.")



def chatbot():

    running = True

    while running:

        user_input = input("\nYou: ")

        user_input = user_input.lower()

        user_input = user_input.strip()


        if user_input == "hello":
            greeting()


        elif user_input == "hi":
            greeting()


        elif user_input == "hey":
            greeting()

        elif user_input == "how are you":
            how_are_you()


        elif user_input == "i am fine":
            good_mood()


        elif user_input == "i am good":
            good_mood()


        elif user_input == "i'm good":
            good_mood()


        elif user_input == "i am sad":
            bad_mood()


        elif user_input == "i am not good":
            bad_mood()


        elif user_input == "what is your name":
            ask_name()


        elif user_input == "who are you":
            introduce()


        elif user_input == "what can you do":
            print("Bot: I can answer simple predefined questions.")


        elif user_input == "help":
            help_user()


        elif user_input == "thank you":
            thank_you()


        elif user_input == "thanks":
            thank_you()


        elif user_input == "good morning":
            good_morning()

        elif user_input == "good afternoon":
            good_afternoon()


        elif user_input == "good evening":
            good_evening()


        elif user_input == "study":
            study()

        elif user_input == "python":
            python()


        elif user_input == "ai":
            artificial_intelligence()


        elif user_input == "data science":
            data_science()


        elif user_input == "chatbot":
            chatbot_info()


        elif user_input == "weather":
            weather()


        elif user_input == "joke":
            joke()


        elif user_input == "motivate me":
            motivation()

        elif user_input == "bye":


            print("Bot: Goodbye!")
            print("Bot: Have a nice day!")


            running = False


        elif user_input == "exit":


            print("Bot: Goodbye!")
            running = False


        elif user_input == "quit":


            print("Bot: See you again!")
            running = False


        else:
            unknown()


chatbot()
```
