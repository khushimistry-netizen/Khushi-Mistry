```python
total_sales = df["Sales"].sum()

print("Total Sales:", total_sales)
```

```python
total_profit = df["Profit"].sum()

print("Total Profit:", total_profit)
```

```python
total_quantity = df["Quantity"].sum()

print("Total Quantity:", total_quantity)
```

```python
average_sales = df["Sales"].mean()

print("Average Sales:", average_sales)
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

print(region_sales)
```

```python
highest_sales = df.loc[df["Sales"].idxmax()]

print(highest_sales)
```

```python
highest_profit = df.loc[df["Profit"].idxmax()]

print(highest_profit)
```

```python
sales = np.array(df["Sales"])

print("Maximum Sales:", np.max(sales))
print("Minimum Sales:", np.min(sales))
print("Average Sales:", np.mean(sales))
print("Total Sales:", np.sum(sales))
```

```python
df["Profit Margin"] = (df["Profit"] / df["Sales"]) * 100

print(df[["Order ID", "Sales", "Profit", "Profit Margin"]])
```
